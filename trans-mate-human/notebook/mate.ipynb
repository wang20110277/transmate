## 测试AI代理功能

### 1. 获取天气信息

```python
# 获取天气信息
async def test_weather():
    result = await agent.ainvoke({
        "messages": [{"role": "user", "content": "北京的天气如何"}]
    })
    print(result["messages"][-1].content)

# 运行异步函数
await test_weather()
```

### 2. 执行数学计算

```python
# 执行数学计算
async def test_calculation():
    result = await agent.ainvoke({
        "messages": [{"role": "user", "content": "计算 25 乘以 15 等于多少"}]
    })
    print(result["messages"][-1].content)

# 运行异步函数
await test_calculation()
```

### 3. 搜索功能

```python
# 使用搜索功能
async def test_search():
    result = await agent.ainvoke({
        "messages": [{"role": "user", "content": "搜索人工智能的最新发展"}]
    })
    print(result["messages"][-1].content)

# 运行异步函数
await test_search()
```

### 4. 多轮对话示例

```python
# 多轮对话测试
async def test_multiple_queries():
    queries = [
        {"messages": [{"role": "user", "content": "上海的天气如何"}]},
        {"messages": [{"role": "user", "content": "计算 25 乘以 15 等于多少"}]},
        {"messages": [{"role": "user", "content": "搜索人工智能的最新发展"}]}
    ]
    
    for query in queries:
        result = await agent.ainvoke(query)
        print(result["messages"][-1].content)

# 运行异步函数
await test_multiple_queries()
```

### 5. 自定义工具使用

```python
# 直接使用自定义工具
weather_result = get_weather.invoke({"city": "广州"})
print(weather_result)
```
